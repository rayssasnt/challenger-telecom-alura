# Dataframe único

```python
data_completo = pd.concat([lista , customer,phone,internet,account] , ignore_index=False , axis=1)
data_completo.head(5)
```

```python
data_completo.info()
```

```python
data_completo['Charges.Total'] = pd.to_numeric(data_completo['Charges.Total'], errors='coerce')
```

```python
data_completo['Charges.Total'].fillna(0)
```

# Converter valores de sim e não em 1 e 0

```python
# converter valores de sim e não em 1 e 0
data_completo.columns
```

```python
lista_converter = ['Churn','Partner','MultipleLines','Dependents','PhoneService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport',
                   'StreamingTV', 'StreamingMovies','PaperlessBilling']
```

## Conferindo valores unicos

```python
for col in lista_converter:
    print(f'\nColuna: {col}')
    print(data_completo[col].unique())
```

## Conversões para 0 e 1

* 1 - Ativado
* 0 - Desativado
* 2 - Sem serviço / sem resposta




```python
data_completo[lista_converter] = data_completo[lista_converter].apply(lambda x: x.replace('Yes','0').replace('No','1').replace('No phone service','2').replace('No internet service','2').replace('','2'))
```

```python
data_completo.head(5)
```

## Transformando as colunas em integer

```python
data_completo[lista_converter] = data_completo[lista_converter].apply(pd.to_numeric, errors='coerce').astype('int64')
```

```python
data_completo.info()
```

```python
# renomear colunas
data_completo.rename(columns={'Churn': 'Cliente_ativo', 'tenure': 'meses_contrato','Charges.Monthly':'valor_mensal','Charges.Total': 'valor_total','SeniorCitizen' : 'clientes_65+','PaperlessBilling' : 'fatura_oline'}, inplace=True)
```

# Relatório Telecom X

# Introdução
Este relatório tem como objetivo a análise de evasão da empresa Telecom X , atraves do processo de ETL e análise exploratória para gerar dados que ajudem a identificar possíveis motivos da evasão

# Limpeza e Tratamento de Dados
Foi realizado uma normalização em Json no dataset original para formatar o DataFrame.

Em seguida foi criado um único DataFrame com todas as tabelas , e transformei as strings SIM e NÃO em valores numéricos para facilitar a análise.

Alteração dos tipo de dado da coluna , alterando alguns em integer ou float e renomeando algumas colunas

# Análise Exploratória

- 26.5% do Clientes deixaram a empresa.
- 42.9% dos Clientes Atuais preferem contratos anuais . em contra partida 88% dos antigos clientes preferam contratos mensais.
- 69.4% dos Clientes Antigos usavam fibra óptica como serviço de internet.
- A média de gasto mensal é maior para Clientes que fazem parte da evasão R$ 74.44
-  A média para a evasão é de 18 meses (1 ano e 6 meses)
- 57.3% dos antigos clientes usam cobrança eletrônica

# Conclusão
- A opção de contrato mensal e alto custo mensal levam o cliente a sair da empresa.
- Os clientes que mais cancelam contarto usam o serviço de fibra óptica
- Clientes que pagam de forma eletrônica tendem a cancelar o serviço , enquanto a cobraça automática tem a menor porcentagem

# Recomendações

- Incentivar pagamento automático
- Melhorar o serviço de fibra óptica
- Icentivar o uso de serviços adicionais
- Recomendar planos de maior duração
- Promoções de boas - vindas para novos clientes
